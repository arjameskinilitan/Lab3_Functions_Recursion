```python
import pandas as pd
import numpy as np

# 1. Configuration & Identification
student_id = 2017
last_three_digits = 17  # Last three digits: 017
filename = 'programming_languages_popularity.csv'

# 2. Load and Prepare Data
df_raw = pd.read_csv(filename)
date_col = df_raw.columns[0]
df_raw[date_col] = pd.to_datetime(df_raw[date_col])

# 3. Determine Assigned Language
# Sort language columns alphabetically (excluding the date column)
languages = sorted([col for col in df_raw.columns if col != date_col])
num_languages = len(languages)
assigned_index = last_three_digits % num_languages
assigned_language = languages[assigned_index]

# 4. Data Extraction
df = df_raw[[date_col, assigned_language]].copy()
df.columns = ['Month', 'Popularity']

# 5. Calculations: Growth Rate and Rolling Stats
df['Growth_Rate'] = df['Popularity'].pct_change() * 100
df['Moving_Avg'] = df['Popularity'].rolling(window=12).mean()
df['Moving_STD'] = df['Popularity'].rolling(window=12).std()

# 6. Classify Growth Phases (Using NumPy Conditional Logic)
conditions = [
    (df['Growth_Rate'] > 5),
    (df['Growth_Rate'] < -5),
    (df['Growth_Rate'].notnull())
]
choices = ['Growth', 'Decline', 'Stable']
df['Phase'] = np.select(conditions, choices, default="N/A")

# 7. Statistical Summary & Key Metrics
summary = df['Popularity'].describe()
initial_pop = df['Popularity'].iloc[0]
final_pop = df['Popularity'].iloc[-1]
overall_growth = ((final_pop - initial_pop) / initial_pop) * 100
phase_counts = df['Phase'].value_counts()

# Output Results
print(f"Assigned Language: {assigned_language}")
print("-" * 30)
print("\nMain Analysis DataFrame (First 5 Rows):")
print(df.head())

print("\nStatistical Summary Table:")
print(summary)

print("\nGrowth Phase Counts:")
print(phase_counts)

print("\nKey Performance Metrics:")
print(f"- Initial Popularity: {initial_pop:.2f}")
print(f"- Final Popularity: {final_pop:.2f}")
print(f"- Overall Growth Percentage: {overall_growth:.2f}%")
print(f"- Periods: {len(df)}")
```

```python
import pandas as pd
import numpy as np

# 1. Identification and Configuration
student_id = "TUPM-24-2017" #
last_three_digits = 17 # Derived from student ID
filename = 'Most Popular Programming Languages.csv' # Verified from environment

try:
    # 2. Data Loading and Cleaning
    df_raw = pd.read_csv(filename)
    date_col = df_raw.columns[0]
    df_raw[date_col] = pd.to_datetime(df_raw[date_col]) # Convert to datetime

    # 3. Language Assignment (Sorted Alphabetically)
    languages = sorted([col for col in df_raw.columns if col != date_col])
    assigned_index = last_three_digits % len(languages)
    assigned_language = languages[assigned_index]

    # 4. Core Calculations
    df = df_raw[[date_col, assigned_language]].copy()
    df.columns = ['Month', 'Popularity']
    
    # Growth Rate Metrics
    df['Growth_Rate'] = df['Popularity'].pct_change() * 100 #
    df['Moving_Avg'] = df['Popularity'].rolling(window=6).mean() # 6-month smoothing
    df['Moving_STD'] = df['Popularity'].rolling(window=6).std() #

    # Statistical Tendency
    growth_mean = np.mean(df['Growth_Rate'].dropna()) #
    growth_std = np.std(df['Growth_Rate'].dropna()) #

    # 5. Lifecycle Rules Definition
    # Applying rules using NumPy conditional logic
    conditions = [
        (df['Growth_Rate'] < 0) & (df['Growth_Rate'] < (-1 * growth_std)), # Decline
        (np.abs(df['Growth_Rate']) <= 1.0),                               # Maturity
        (df['Growth_Rate'] > growth_mean),                                # Growth
        (df['Growth_Rate'] > 0) & (df['Growth_Rate'] <= growth_mean)      # Introduction
    ]
    choices = ['Decline', 'Maturity', 'Growth', 'Introduction']
    df['Lifecycle_Phase'] = np.select(conditions, choices, default="Stable/Indeterminate")

    # 6. Summary and Interpretive Metrics
    phase_counts = df['Lifecycle_Phase'].value_counts()
    dominant_stage = phase_counts.idxmax()
    
    # Output Results
    print(f"Assigned Language: {assigned_language}")
    print("-" * 50)
    print("\nLifecycle Classification Table (Sample):")
    print(df[['Month', 'Popularity', 'Growth_Rate', 'Moving_Avg', 'Lifecycle_Phase']].tail(10))

    print("\n--- Summary Distribution ---")
    print(phase_counts)
    
    print(f"\nInterpretive Insight: {assigned_language} is primarily in the {dominant_stage} phase.")
    print("This implies a consistent global trajectory relative to historical volatility.")

except FileNotFoundError:
    print(f"Error: '{filename}' not found. Ensure the file is in your project directory.")
```

```python
import pandas as pd
import numpy as np

# 1. Configuration & Identification
student_id = "TUPM-24-2017"
last_three_digits = 17 
filename = 'Most Popular Programming Languages.csv'

try:
    # 2. Load and Prepare Data
    df_raw = pd.read_csv(filename)
    date_col = df_raw.columns[0]
    df_raw[date_col] = pd.to_datetime(df_raw[date_col])

    # 3. Language Selection Rule
    # Alphabetical list of languages excluding the date column
    languages = sorted([col for col in df_raw.columns if col != date_col])
    
    # First language based on student ID index
    idx1 = last_three_digits % len(languages)
    lang_a_name = languages[idx1]
    
    # Second language is the next one alphabetically (wraps to start if at end)
    idx2 = (idx1 + 1) % len(languages)
    lang_b_name = languages[idx2]

    # 4. Comparative Calculations
    data_a = df_raw[lang_a_name]
    data_b = df_raw[lang_b_name]

    # Central Tendencies
    stats = {
        'Mean_A': data_a.mean(),
        'Mean_B': data_b.mean(),
        'Std_A': data_a.std(),
        'Std_B': data_b.std(),
        'Median_A': data_a.median(),
        'Median_B': data_b.median()
    }

    # Correlation Coefficient
    correlation = np.corrcoef(data_a, data_b)[0, 1]

    # Dominance Ratio
    months_a_leads = np.sum(data_a > data_b)
    total_months = len(df_raw)
    dominance_ratio = (months_a_leads / total_months) * 100

    # Relative Performance Index (RPI)
    # Simple RPI defined as the average ratio of popularity over time
    rpi = (data_a / data_b).mean()

    # Identify Crossover Points (Months when A surpasses B)
    # We find where A > B was false in previous month but true now
    crossovers = df_raw[date_col][(data_a > data_b) & (data_a.shift(1) <= data_b.shift(1))]

    # 5. Output Summary Table
    summary_df = pd.DataFrame({
        'Metric': ['Mean', 'Std Dev', 'Correlation', 'Dominance Ratio (%)', 'RPI'],
        lang_a_name: [stats['Mean_A'], stats['Std_A'], correlation, dominance_ratio, rpi],
        lang_b_name: [stats['Mean_B'], stats['Std_B'], correlation, "N/A", "N/A"]
    })

    print(f"Assigned Languages: {lang_a_name} vs {lang_b_name}")
    print("-" * 60)
    print("\nComparative Statistics Summary:")
    print(summary_df.to_string(index=False))

    if not crossovers.empty:
        print("\nSignificant Crossover Points (A surpassed B):")
        print(crossovers.head().to_string(index=False))
    else:
        print(f"\nNo crossover points detected: {lang_a_name} or {lang_b_name} maintained dominance.")

    # 6. Interpretation
    if stats['Std_A'] < stats['Std_B']:
        print(f"\nInterpretation: {lang_a_name} demonstrates more consistent growth with lower variability.")
    else:
        print(f"\nInterpretation: {lang_b_name} demonstrates more resilient stability or consistent trends.")

except FileNotFoundError:
    print(f"Error: The file '{filename}' was not found in your directory.")
except Exception as e:
    print(f"An error occurred during analysis: {e}")
```
