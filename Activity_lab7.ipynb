```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Identification and Parameters
student_name = "Arjames E Kinilitan"
student_id = "2017"
color_bar = "skyblue"
color_line = "salmon"

# 2. Prepare Data
# Ensure you have a 'release_year' column; if not, create it from 'release_date'
# df['release_year'] = pd.to_datetime(df['release_date']).dt.year

# 3. Calculations
# Group by year and count songs
yearly_counts = df.groupby('release_year').size().reset_index(name='song_count')
yearly_counts = yearly_counts.sort_values('release_year')

# Calculate cumulative song count
yearly_counts['cumulative_count'] = yearly_counts['song_count'].cumsum()

# 4. Visualization: Combined Bar + Line Chart
fig, ax1 = plt.subplots(figsize=(10, 6))

# Plot Yearly Counts as Bars
ax1.bar(yearly_counts['release_year'], yearly_counts['song_count'], color=color_bar, alpha=0.7, label='Yearly Count')
ax1.set_xlabel('Year')
ax1.set_ylabel('Number of Songs Released', color=color_bar)
ax1.tick_params(axis='y', labelcolor=color_bar)

# Create a secondary y-axis for the Cumulative Total
ax2 = ax1.twinx()
ax2.plot(yearly_counts['release_year'], yearly_counts['cumulative_count'], color=color_line, marker='o', linewidth=2, label='Cumulative Total')
ax2.set_ylabel('Cumulative Song Count', color=color_line)
ax2.tick_params(axis='y', labelcolor=color_line)

# Title and Layout
plt.title(f"{student_name} ({student_id})\nSong Release Trends: Yearly vs. Cumulative")
fig.tight_layout()

# Adding a combined legend
lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines + lines2, labels + labels2, loc='upper left')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.animation as animation

# 1. Identification and Parameters
student_name = "Arjames E Kinilitan"
student_id = "2017"
cmap_color = "magma"  # Student-specific color map

# 2. Setup Figure
fig, ax = plt.subplots(figsize=(8, 6))
years = sorted(df['release_year'].unique())

# 3. Animation Update Function
def update(year):
    ax.clear()
    # Filter data for the specific year
    data_subset = df[df['release_year'] == year]
    
    # Generate Hexbin Map
    hb = ax.hexbin(data_subset['duration_min'], 
                   data_subset['popularity'], 
                   gridsize=15, 
                   cmap=cmap_color, 
                   mincnt=1)
    
    # Formatting and Personalization
    ax.set_title(f"{student_name} ({student_id})\nDensity Map: Duration vs Popularity ({year})")
    ax.set_xlabel('Duration (min)')
    ax.set_ylabel('Popularity')
    
    # Keep axes limits constant for consistent animation
    ax.set_xlim(df['duration_min'].min() - 0.5, df['duration_min'].max() + 0.5)
    ax.set_ylim(0, 105)
    return hb,

# 4. Create and Save Animation
ani = animation.FuncAnimation(fig, update, frames=years, interval=1000)
# To save in VS Code, ensure you have 'pillow' installed: pip install pillow
ani.save('density_evolution.gif', writer='pillow')
plt.show()
```
